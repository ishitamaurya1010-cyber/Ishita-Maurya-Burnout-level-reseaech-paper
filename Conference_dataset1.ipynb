```python

import pandas as pd
if 'job_role' not in df.columns:
    # Load the full original dataset to retrieve the 'job_role' column
    original_full_df = pd.read_csv("archive (1).zip")
    # Assuming the row order hasn't changed, add the column back to df.
    df['job_role'] = original_full_df['job_role']

for col in text_columns:
    if col != 'job_role': # job_role was already tokenized
        df[col + '_tokens'] = df[col].astype(str).apply(word_tokenize)

df.head()
```

```python
#Lemmezation
from nltk.stem import WordNetLemmatizer
import re
import nltk
nltk.download('stopwords')
nltk.download('wordnet') # Download wordnet for lemmatization
from nltk.corpus import stopwords
lemmatizer = WordNetLemmatizer()

def clean_and_lemmatize(text):
  text=text.lower()
  text=re.sub(r'[^a-zA-Z0-9\s]','',text)
  tokens = text.split()
  tokens = [word for word in tokens if word not in stopwords.words('english')]
  tokens = [lemmatizer.lemmatize(word) for word in tokens]
  return " ".join(tokens)

df['cleaned_review'] = df['job_role'].apply(clean_and_lemmatize)
```

```python
for col in text_columns:
  # Create a new column for each lemmatized text column
  df[col + '_lemmatized'] = df[col].astype(str).apply(clean_and_lemmatize)

# Display the head of the DataFrame to show the new lemmatized columns
df.head()
```

```python
#Vectorization
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Identify the lemmatized columns created in the previous step
#    Assuming 'text_columns' still holds the original column names, we need the lemmatized versions.
lemmatized_cols = [col + '_lemmatized' for col in text_columns]

# 2. Combine the text from all lemmatized columns into a single 'combined_text' column
#    Fill NaN with empty string to avoid errors during concatenation.
df['combined_text'] = df[lemmatized_cols].fillna('').agg(' '.join, axis=1)

# 3. Initialize TfidfVectorizer
vectorizer = TfidfVectorizer(stop_words='english')

# 4. Fit and transform the combined text
#    This will return a single sparse matrix X representing all features from all combined text.
X = vectorizer.fit_transform(df['combined_text'])

# 5. Create a DataFrame from the sparse matrix
tfidf_pdata = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())

# 6. Save and print head
tfidf_pdata.to_csv("vectorized_textdata.csv", index=False)
print(tfidf_pdata.head())
```

```python

from sklearn.preprocessing import LabelEncoder
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, MaxPooling1D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split
import numpy as np

# Define early_stopping callback within this cell to ensure it's always available
early_stopping=tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    min_delta=0.0001,
    patience=20,
    verbose=1,
    mode="auto",
    baseline=None,
    restore_best_weights=False,
)

# 1. Prepare X data (padded sequences) for the CNN
# Re-initialize tokenizer and fit on the combined text from the DataFrame
tokenizer = Tokenizer(num_words=None, oov_token="<unk>")
tokenizer.fit_on_texts(df['combined_text'])

# Convert text to sequences
X_sequences = tokenizer.texts_to_sequences(df['combined_text'])

# Determine vocab_size and max_len here to ensure they are defined
vocab_size = len(tokenizer.word_index) + 1
max_len = max(len(s) for s in X_sequences) if X_sequences else 0 # Handle empty sequences case

# Pad sequences
X_padded = pad_sequences(X_sequences, maxlen=max_len, padding='post')

# 2. Prepare y data (encoded labels) and perform train-test split for CNN
le = LabelEncoder()
y_encoded = le.fit_transform(original_full_df['burnout_level'])

# Perform train-test split on the padded sequences and encoded labels
X_train_cnn, X_test_cnn, y_train_encoded, y_test_encoded = train_test_split(
    X_padded, y_encoded, test_size=0.3, random_state=42
)

# 3. Re-initialize and build the CNN model with a compatible architecture
# This will overwrite the previously problematic 'model' object.
model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=100, input_length=max_len))

# Revised CNN layers for max_len = 6
# (Output after Embedding: (None, 6, 100))
model.add(Conv1D(filters=64, kernel_size=2, activation='relu'))
model.add(Dropout(0.5)) # Added Dropout layer
# (Output after first Conv1D: (None, 5, 64))
model.add(MaxPooling1D(pool_size=2))
# (Output after MaxPooling1D: (None, 2, 64))
model.add(Conv1D(filters=128, kernel_size=2, activation='relu'))
model.add(Dropout(0.5)) # Added Dropout layer
# (Output after second Conv1D: (None, 1, 128))
model.add(Flatten())
# (Output after Flatten: (None, 128))

# Adding dense layers
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5)) # Added Dropout layer
model.add(Dense(64, activation='relu'))
model.add(Dropout(0.5)) # Added Dropout layer
model.add(Dense(32, activation='relu'))
model.add(Dropout(0.5)) # Added Dropout layer
model.add(Dense(3, activation='softmax')) # Assuming 3 classes for burnout_level

# 4. Compile the new model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# 5. Fit the model with the correct data
model_history = model.fit(X_train_cnn, y_train_encoded, validation_split=0.33, batch_size=10, epochs=1000, callbacks=[early_stopping])
```

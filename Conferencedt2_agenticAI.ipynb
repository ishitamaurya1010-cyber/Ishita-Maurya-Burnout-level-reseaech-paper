```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df=pd.read_csv("Mental Distress Dataset-original.csv")
```

```python
df.head()
```

```python
df['text']=df['text'].str.lower()
```

```python
df.head()
```

```python
#Converting columns contatining text data to numeric data
```

```python
import string
exclude=string.punctuation
```

```python
def remove_punc(text):
  # Ensure text is treated as a string to avoid AttributeError with floats (NaNs)
  text = str(text)
  return text.translate(str.maketrans('','',exclude))
df['text']=df['text'].apply(remove_punc)
```

```python
df.head()
```

```python
#Bert tokenizer
from transformers import BertTokenizer, BertModel
import torch

# 1. Load tokenizer & model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

# 2. Preprocess text
inputs = tokenizer(df['text'].tolist(),batch_size=8, return_tensors="pt", padding=True, truncation=True)

# 3. Get embeddings
outputs = model(**inputs)
embedding = outputs.last_hidden_state[:,0,:]  # CLS token embedding
```

```python
#Bert embedding
import torch
from transformers import AutoTokenizer, AutoModel

# 1. Load the pre-trained BERT tokenizer and model
model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

# 2. Define the input text
text = "Python code for BERT embeddings is highly efficient."

# 3. Tokenize input and convert to PyTorch tensors
inputs = tokenizer(
    text,
    return_tensors="pt",
    padding=True,
    truncation=True,
    add_special_tokens=True
)

# 4. Generate embeddings without tracking gradients (inference mode)
with torch.no_grad():
    outputs = model(**inputs)

# 5. Extract specific types of embeddings

# A. Word/Token-Level Embeddings
# Shape: [batch_size, sequence_length, hidden_dimension] -> (1, sequence_length, 768)
word_embeddings = outputs.last_hidden_state

# B. Sentence-Level Embedding Option 1: [CLS] Token
# The first token ([CLS]) represents the aggregated sentence meaning
sentence_embedding_cls = word_embeddings[0, 0, :]

# C. Sentence-Level Embedding Option 2: Mean Pooling (Recommended)
# Average out all token embeddings across the sequence length dimension
sentence_embedding_mean = word_embeddings.mean(dim=1).squeeze()

# Print results and tensor shapes
print(f"Token Embeddings Shape: {word_embeddings.shape}")
print(f"CLS Sentence Embedding Shape: {sentence_embedding_cls.shape}")
print(f"Mean Pooled Sentence Embedding Shape: {sentence_embedding_mean.shape}")
```
